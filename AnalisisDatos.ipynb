# Taller Práctico de Análisis de Datos con Pandas

## Flota de Transporte y Envíos Logísticos

```python
import pandas as pd
```

## 1. Lectura y estructura base

```python
# Importar el DataFrame desde el archivo envios_transporte.csv
df_envios = pd.read_csv("envios_transporte.csv")
df_envios
```

## 2. Proyección y separación de series

```python
# Extrae la columna Tipo_Vehiculo como una Serie de pandas
tipo_vehiculo = df_envios["Tipo_Vehiculo"]
tipo_vehiculo, type(tipo_vehiculo)
```

```python
# DataFrame reducido que presente las columnas Ruta_Origen, Tipo_Carga y Distancia_KM
df_reducido = df_envios[["Ruta_Origen", "Tipo_Carga", "Distancia_KM"]]
df_reducido
```

## 3. Acceso por etiqueta (loc)

```python
# Mostrar la informacion completa del despacho registrado en la fila con índice 99 utilizando el método .loc[]
despacho_99 = df_envios.loc[99]
despacho_99
```

## 4. Limpieza de registros duplicados

```python
# Eliminar todos los registros duplicados
print("Filas antes:", len(df_envios))
print("Duplicados encontrados:", df_envios.duplicated().sum())

df_sin_duplicados = df_envios.drop_duplicates()

print("Filas después:", len(df_sin_duplicados))

df_sin_duplicados
```

## 5. Filtrado por condición simple

```python
# Filtrado por condición simple: despachos con más de 15 toneladas
df_carga_pesada = df_sin_duplicados[df_sin_duplicados["Peso_Toneladas"] > 15.0]

print("Despachos con más de 15 toneladas:", len(df_carga_pesada))

df_carga_pesada
```

## 6. Filtrado categórico con .isin()

```python
# Filtrar por tipo de vehiculo (isin)
df_vehiculos = df_sin_duplicados[df_sin_duplicados["Tipo_Vehiculo"].isin(["Tractomula", "Camión Mediano"])]
df_vehiculos
```

## 7. Filtrado de rangos numéricos (between)

```python
# Filtrar las rutas con Distancia_KM entre 100.0 y 250.0
df_distancia = df_sin_duplicados[df_sin_duplicados["Distancia_KM"].between(100.0, 250.0)]
df_distancia
```

## 8. Consultas con el método .query()

```python
# Envios originados en Pereira con Estado_Entrega Retrasado
df_pereira_retrasado = df_sin_duplicados.query('Ruta_Origen == "Pereira" and Estado_Entrega == "Retrasado"')
df_pereira_retrasado
```

## 9. Filtrado lógico compuesto y cadenas

```python
# Despachos A Tiempo y cuyo Tipo_Carga comience con la letra "E"
df_a_tiempo_e = df_sin_duplicados[
    (df_sin_duplicados["Estado_Entrega"] == "A Tiempo") &
    (df_sin_duplicados["Tipo_Carga"].str.startswith("E"))
]
df_a_tiempo_e
```

## 10. Tratamiento e imputación de nulos

```python
# Rellenar valores faltantes
df_faltantes = df_sin_duplicados.fillna(
    {
        "Distancia_KM": df_sin_duplicados["Distancia_KM"].mean(),
        "Combustible_Galones": 0.0
    }
)

df_faltantes
```

## 11. Columnas calculadas con .apply()

```python
#Transformacion de datos
def obtener_categoria_carga(peso):
    if peso < 5.0:
        return "Ligero"
    elif peso <= 15.0:
        return "Intermedio"
    else:
        return "Pesado"

df_faltantes["Categoria_Carga"] = df_faltantes["Peso_Toneladas"].apply(obtener_categoria_carga)
df_faltantes
```

## 12. Renombrado y reordenamiento

```python
# Renombrar columnas
df_renombrado = df_faltantes.rename(
    columns={
        "Ruta_Origen": "Origen",
        "Distancia_KM": "Recorrido_KM"
    }
)
df_renombrado
```

```python
# Ordenar columnas
columnas_ordenadas = ["ID_Envio", "Categoria_Carga", "Origen", "Tipo_Vehiculo", "Tipo_Carga", "Recorrido_KM", "Combustible_Galones", "Peso_Toneladas", "Estado_Entrega"]
df_ordenado = df_renombrado[columnas_ordenadas]
df_ordenado
```

## 13. Agrupación y métricas globales

```python
# Agrupar los datos por origen
grouped = df_ordenado.groupby("Origen")
grouped.groups
```

```python
# Calcular el promedio de Recorrido_KM y la suma de Combustible_Galones
aggregate_data = grouped.agg(
    {
        "Recorrido_KM": "mean",
        "Combustible_Galones": "sum"
    }
)
aggregate_data
```

## 14. Dispersión con función personalizada

```python
# Definir una funcion de agregacion personalizada
def rango(series):
    return series.max() - series.min()

# Agrupar los datos por tipo de vehiculo
grouped_vehiculo = df_ordenado.groupby("Tipo_Vehiculo")

# Aplicamos la funcion agg pero al grupo
aggregate_data_custom = grouped_vehiculo.agg(
    {
        "Peso_Toneladas": rango,
        "Recorrido_KM": rango
    }
)

aggregate_data_custom
```

## 15. Agrupamiento multinivel y consolidación

```python
# Agrupar datos por origen y tipo de vehiculo
grouped_multi = df_ordenado.groupby(["Origen", "Tipo_Vehiculo"])
# Ver en tabla los envios de un grupo (Armenia y Tractomula)
grouped_multi.get_group(("Armenia", "Tractomula"))
```

```python
# Calcular el promedio de Recorrido_KM y la suma de Peso_Toneladas por origen y tipo de vehiculo
aggregate_data_multi = grouped_multi.agg(
    {
        "Recorrido_KM": "mean",
        "Peso_Toneladas": "sum"
    }
)

aggregate_data_multi
```
